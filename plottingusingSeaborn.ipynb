# Relational PLot

```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
tips = sns.load_dataset('tips')
```

```python
#scatter plot-> axes level function
sns.scatterplot(data=tips,x='total_bill',y ='tip',hue='sex',style='time',size='size')
```

```python
# relplot -> figure level -> squre level
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',hue='sex',style='time',size='size')
```

```python
#line plot
gap=px.data.gapminder()
gap
```

```python
temp_df=gap[gap['country']=='India']
```

```python
sns.lineplot(data=temp_df,x='year',y='lifeExp')
```

```python
#using relplot
sns.relplot(data=temp_df,x='year',y='lifeExp',kind='line')
```

```python
# hue -> style
temp_df = gap[gap['country'].isin(['India','Pakistan','China'])]
```

```python
sns.relplot(kind='line',data=temp_df,x='year',y='lifeExp',hue='country')
```

```python
sns.lineplot(data=temp_df,x='year',y='lifeExp',hue='country')
```

```python
temp_df = gap[gap['country'].isin(['India','Brazil','Germany'])]
temp_df
```

```python
sns.relplot(kind='line',data=temp_df,x='year',y='lifeExp',hue='country',style='continent',size='continent')
```

```python
#facet plot -> figure level function -> work with relplot
# it will not work with scatterplot and lineplot
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',col='sex',row='smoker')
```

```python
sns.relplot(data=tips,x='total_bill',y='tip',kind='line',col='sex',row='smoker')
```

```python
# col wrap
sns.relplot(data=gap,x='lifeExp',y='gdpPercap',kind='scatter',col='year',col_wrap=3)
```

# Distribution Plot

```python
#figure level -> displot
#axes level -> histplot -> kdeplot -> rugplot
```

```python
#plotting univariate histogram
sns.histplot(data=tips,x='total_bill')
```

```python
sns.displot(data=tips,x='total_bill',kind='hist')
```

```python
# bins parameter
sns.displot(data=tips,x='total_bill',kind='hist',bins=20)
```

```python
#it's also possible to visulize the distribution of a categorical variable using the logic of a histogram.
#discrete bins are automatic44et for categorical variable
#countplot
sns.displot(data=tips,x='day',kind='hist')
```

```python
# hue parameter 
sns.displot(data=tips,x='tip',kind='hist',hue='sex')
```

```python
#element -> step
sns.displot(data=tips,x='tip',kind='hist',hue='sex',element='step')
```

```python
titanic=sns.load_dataset('titanic')
```

```python
titanic
```

```python
sns.displot(data=titanic,x='age',kind='hist',hue='sex',element='step')
```

```python
#faceting  usnig  col and row  -> not work on histplot function
sns.displot(data=tips,x='tip',kind='hist',col='sex',element='step')
```

```python
# kdeplot
#rather then using discrete bins, a KDE plot smooths the observations with a gaussian kernel ,producing a continuous density estimate
sns.kdeplot(data=tips,x='total_bill')
```

```python
# hue -> fill
sns.displot(data =tips,x = 'total_bill',kind='kde',hue='sex',fill=True)
```

```python
# Rugplot
#plot marginal distribution by drawning ticks along the x and y axes.
#this function is intended to complement other plots by showing the location of individual observation is an unobtrusive way.
sns.kdeplot(data=tips ,x='total_bill')
sns.rugplot(data=tips ,x='total_bill')
```

```python
# bivariate histogram
# A bivariate histogram bins the data within rectangles that tile the plot
# and then shows the count of observation within each rectangle with the fill color
sns.histplot(data=tips,x='total_bill',y='tip')
```

```python
# bivariate kdeplot
#a bivariate kde plot smoothes the (x,y) observation with a 2D gaussian 
sns.kdeplot(data=tips,x='total_bill',y='tip')
```

# Matrix Plot
* Heatmap
* Clustermap

```python
# Heat map
#plot rectangle data as data as a color-encoded matrix
temp_df=gap.pivot(index='country',columns='year',values='lifeExp')

# axes level function
plt.figure(figsize=(15,10))
sns.heatmap(temp_df)
```

```python
# annot
temp_df=gap[gap['continent']=='Europe'].pivot(index='country',columns='year',values='lifeExp')
```

```python
plt.figure(figsize=(15,15))
sns.heatmap(temp_df,annot=True,linewidth=2,cmap='summer')
```

```python
# Clustermap
# plot a matrix dataset as a hierarchically-clustered heatmap
# this function requires scipy to be available
iris=px.data.iris()
```

```python
iris.head()
```

```python
sns.clustermap(iris.iloc[:,[0,1,2,3]])
```

# Plotting Pairwise Relationship (PairGrid VS Pairplot)

```python
iris
```

```python
sns.pairplot(iris,hue='species')
```

```python
g = sns.PairGrid(data=iris)
# g.map
g.map(sns.scatterplot)
```

```python
#map_diag -> map_offdiag
g = sns.PairGrid(data=iris,hue='species')
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)
```

```python
g = sns.PairGrid(data=iris,hue='species')
g.map_diag(sns.violinplot)
g.map_offdiag(sns.scatterplot)
```

```python
#map_diag->map_upper->map_lower
g = sns.PairGrid(data=iris,hue='species',vars=['sepal_width','petal_width'])
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
```
