```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# plotting a simple function
year = [2025,2026, 2027, 2028, 2029, 2030, 2031]
price = [2000, 2200, 2500, 3000, 3500, 3700, 3900]
plt.plot(year, price)
```

```python
#from a pandas dataframe
batsman = pd.read_csv('sharma-kohli.csv')
batsman
plt.plot(batsman['index'],batsman['V Kohli'])
```

```python
# plotting multiple plots
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
#LABEL TITLE
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
plt.title('rohit sharma or virat kohli career comparison')
plt.ylabel('Runs Scored')
plt.xlabel('season')
```

```python
#color(hex) and line(width and style) and marker(size)
plt.plot(batsman['index'],batsman['V Kohli'],color = 'red')
plt.plot(batsman['index'],batsman['RG Sharma'],color = 'blue')
plt.title('rohit sharma or virat kohli career comparison')
plt.ylabel('Runs Scored')
plt.xlabel('season')
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color = 'red',linestyle = 'dashdot',linewidth = 4)
plt.plot(batsman['index'],batsman['RG Sharma'],color = 'blue',linestyle ='dashdot',linewidth = 1)
plt.title('rohit sharma or virat kohli career comparison')
plt.ylabel('Runs Scored')
plt.xlabel('season')
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color = 'red',linestyle = 'dashdot',linewidth = 4, marker = 'D')
plt.plot(batsman['index'],batsman['RG Sharma'],color = 'blue',linestyle ='dashdot',linewidth = 1,marker = 'o')
plt.title('rohit sharma or virat kohli career comparison')
plt.ylabel('Runs Scored')
plt.xlabel('season')
```

```python
#legend -> location
plt.plot(batsman['index'],batsman['V Kohli'],color = 'red',linestyle = 'dashdot',linewidth = 4, marker = 'D', label = 'virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color = 'blue',linestyle ='dashdot',linewidth = 1,marker = 'o',label = 'rohit')
plt.title('rohit sharma or virat kohli career comparison')
plt.ylabel('Runs Scored')
plt.xlabel('season')
plt.legend()
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color = 'red',linewidth = 2, marker = 'D', label = 'virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color = 'blue',linestyle ='dashdot',linewidth = 1,marker = 'o',label = 'rohit')
plt.title('rohit sharma or virat kohli career comparison')
plt.ylabel('Runs Scored')
plt.xlabel('season')
plt.legend(loc = 'upper right')
plt.grid()
```

# Bar chart 

```python
#simple bar chart
children = [10,15,14,16,5]
colours = ['red' ,'blue', 'green', 'skyblue','black']
plt.bar(colours,children)
```

```python
children = [10,15,14,16,5]
colours = ['red' ,'blue', 'green', 'skyblue','black']
plt.bar(colours,children,color = 'black')
```

```python
df1 = pd.read_csv('batsman_season_record.csv')
df1
```

```python
plt.bar(np.arange(df1.shape[0]) - 0.2, df1['2015'],width=0.2)
plt.bar(np.arange(df1.shape[0]) , df1['2016'],width=0.2)
plt.bar(np.arange(df1.shape[0]) + 0.2, df1['2017'],width=0.2)
plt.xticks(np.arange(df1.shape[0]), df1['batsman'])
plt.show()
```

```python
np.arange(df1.shape[0])
```

```python
children = [10,15,14,16,5]
colours = ['red redredredredred' ,'blubluebluebluebluee', 'greenblueblueblueblue', 'skyblueblueblueblueblue','blackblueblueblue']
plt.bar(colours,children)
plt.xticks(rotation='vertical')
plt.show()
```

```python
df1
```

```python
# stacked bar chart
plt.bar(df1['batsman'],df1['2017'],label='2017')
plt.bar(df1['batsman'],df1['2016'],bottom=df1['2017'],label='2016')
plt.bar(df1['batsman'],df1['2015'],bottom=(df1['2016']+df1['2017']),label='2015')
plt.legend()
plt.show()
```
